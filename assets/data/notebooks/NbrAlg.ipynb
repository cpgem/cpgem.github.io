# Nombres algébriques
On rappelle qu'un nombre complexe $z$ est dit algébrique s'il existe au moins un polynôme non nul $P\in\mathbb Q[X]$ tel que $P(z)=0$. $P$ est alors dit un polynôme annulateur de $z$.

Soit dans la suite $z$ un nombre algébrique. L'ensemble 
$$I_z=\{P\in\mathbb Q[X]\mid P(z)=0\}$$
est un idéal non nul de $\mathbb Q[X]$. Il existe donc un unique polynôme unitaire $\pi_z$ tel que 
$$I_z=\pi_z\mathbb Q[X]$$
Ainsi $\pi_z$ est le polynôme unitaire de degré minimal qui annule $z$. Il est immédiat que ce polynôme est irréductible dans $\mathbb Q[X]$. On pose en outre 
$$\mathbb Q[z]=\{P(z) \mid P\in\mathbb Q[X]\}$$
et on démontre que $\mathbb Q[z]$ est une $\mathbb Q$-algèbre de dimension égale à $\deg\pi_z$. Un élément de $Q(z)$ de $\mathbb Q[z]$ est inversible dans $\mathbb Q[z]$ si et seulement si $Q$ et premier avec $\pi_z$ et grâce à l'irréductibilité de $\pi_z$ on en déduit qu'en fait tout élément non nul de $\mathbb Q[z]$ est inversible dans $\mathbb Q[z]$. Résumons 
$$\text{$\mathbb Q[z]$ est un corps de dimension $d=\deg\pi_z$ sur $\mathbb Q$}$$
---
Considérons maintenant un polynôme non constant $P$ de $\mathbb Q[X]$. On appelle corps de décomposition de $P$ sur $\mathbb Q$ le plus petit sous-corps de $\mathbb C$ qui contient toutes les racines de $P$. Un tel corps existe,  si $z_1,z_2,\ldots,z_p$ sont les racines distinctes de $P$ alors on peut le construire par récurrence en posant pour tout $k$ entre $2$ et $p$ :
$$\mathbb Q[z_1,z_2,\ldots,z_k]:=\mathbb Q[z_1,\ldots,z_{k-1}][z_k]$$
Le corps de décomposition de $P$ est alors $\mathbb Q[z_1,z_2,\ldots,z_p]$.

La théorie des corps fournit un résultat qui affirme pour tout polynôme $P$ de $\mathbb Q[X]$ irréductible sur $\mathbb Q$ de racines $z_1,z_2,\ldots,z_p$ (distinctes) dans $\mathbb C$ il existe $\omega\in\mathbb C$ tel que 
$$ \mathbb Q[z_1,z_2,\ldots,z_p]=\mathbb Q[\omega] $$
Le nombre $\omega$ est alors dit un nombre primitif du corps de décompositions de $P$. Noter qu'en particulier toutes les racines de $P$ sont des polynômes en $\omega$.

```python
from sympy import *
x,y,z=symbols('x y z')
```

```python
minpoly(sqrt(2))
```

```python
minpoly(sqrt(2)+sqrt(3))
```

```python
P=x**3-3*x**2+1
Q=x**4-3*x**3+1
```

```python
factor(P)
```

```python
roots(P)
```

```python
alpha=AlgebraicNumber(RootOf(P,0),alias='alpha')
```

```python
factor(P,extension=alpha)
```

```python
alpha**(12)-4*alpha**2+1
```

```python
poly(alpha**(12)-4*alpha**2+1,y,extension=alpha).as_expr()
```

```python
factor(Q)
```

```python
beta=symbols('beta:4')
beta=[AlgebraicNumber(RootOf(Q,k),alias=beta[k]) for k in range(4)]
```

```python
R,L=primitive_element((beta[0],beta[1]))
```

```python
display(R,L)
```

```python
omega=AlgebraicNumber(RootOf(R,0),alias='omega')
```
