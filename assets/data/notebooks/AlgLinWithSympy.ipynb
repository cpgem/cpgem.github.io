## Symboles et expressions 

### Declaration de symboles

La bibliothèque `sympy` permet de faire ce qu'on appelle du calcul symbolique, par opposition à `numpy/scipy` qui fait du calcul numérique. Concrètement, elle est capable de manipuler des symboles qui ont un sens mathématique, comme des noms de variables ou de fonctions, comme on le ferait à la main. Avant d'utiliser un symbole, il faut le déclarer en tant que tel sinon Python se plaindra d'une variable inconnue. L'une des fonctions `sympy` pour le faire est `symbols`. 

```python
x+y
```

Dans le code précédent, Python se plaint de variables non définies. Ce qui est normal puisque les noms 'x' et 'y' n'ont fait l'objet d'aucune affectation. Mais alors comment faire du calcul symbolique dans ces conditions ? Comment définir un polynôme avec une ou plusieurs indéterminé par exemple ?   

```python
symbols('x')+symbols('y') # ou S('x')+S('y') en plus court 
```

```python
symbols('x')**2+2*symbols('x')+1
```

```python
factor(_) # dans une session interactive, _ fait référence au résultat de l'instruction précédente. 
```

Il serait trop lourd de toujours utiliser des noms de variables de cette manière. En fait il suffit d'affecter à `x` le résultat de `symbols('x')`pour pouvoir utiliser ensuite `x`exactement comme si on fait appel à `symbols('x')`. 

```python
x=symbols('x')
factor(x**3-3*x**2+3*x-1)
```

Rien n'oblige que le nom de la variable soit le même que le nom du «symbole»

```python
om=symbols('omega') # utiliser le nom de la variable et non celui du symbole dans les expressions qui utilisent le symbole
expand((om+1)*(om-1)) 
```

On peut procéder à une affectation par lot (pour déclarer tous les symboles de la session par exemple)

```python
x,y,z,t=symbols('x y z t') # declare les noms x, y, z et t comme des variables symboliques. 
```

Et même utiliser des «indices»

```python
symbols('m1:5') # crée un tuple de symboles
```

```python
m=symbols('m:10') ; m # m sera un tuple, ce qui permet de créer en une seule instruction toute une famille de symboles qui seront par exemple utilisés dans une équation, un système linéaire,... 
```

```python
m[3] # la variable m3. Utiliser m3 provoquerait une erreur de variable non définie
```

On peut faire des déclarations sur la natures sur les symboles. Déclarations qui seront ensuite utilisées dans les opérations de transformation d'une éventuelle expression qui utilise les symboles.

```python
sqrt(x**2)
```

```python
x=symbols('x',positive=True) # declare 'x' comme symbolisant un réel positif
```

```python
sqrt(x**2)
```

Les déclarations possibles :
+ `integer=(True|False)`
+ `positive=(True|False)`
+ `real=(True|False)`
+ `imaginary=(True|False)`
+ `complex=(True|False)`

### Représentation des nombres

Les fractions (nombres rationnels) ont un comportement particulier 

```python
sin(1/2*x)
```

Ici SymPy a transformé la farction $1/2$ en un flottant. En fait il n'en est pas responsable. C'est le moteur Python sous-jascent qui a effectué le calcul avant de fournir le résultat aux fonctions Sympy. Si on tient à conserver la fraction $1/2$, il faudra changer sa signification avec la fonction `Rational`

```python
sin(Rational(1/2)*x)
```

```python
sin(Rational(1)/2*x) # ici Rational(1) empêche Python de «voir» le nombre 1/2. 
```

La fonction `S` peut être aussi utilisée dans de tels situations

```python
sin(S(1)/2*x)
```

Certains nombres sont au contraire directement représentés par des symboles et il faut à l'inverse forcer leurs évaluation en des nombres flottants avec la méthode `evalf` si on veut les remplacer par des valeurs approchées. Les symboles `pi`, `E` et `I` en font partie.

```python
print(f"pi = {pi}, sin(pi) = {sin(pi)}, cos(pi) = {cos(pi)}, exp(I*pi) = {exp(I*pi)}, exp(I*pi/2) = {exp(I*pi/2)}")
```

```python
print('pi ~ ',pi.evalf(50), 'E ~ ',E.evalf(50))
```

```python
acos(0) # arccos(0)
```

## Polynômes avec SymPy

Un polynôme peut être créé comme une simple expression utilisant des «symboles» déclarés auparavant. Une expression peut avoir plusieurs indéterminées.  

```python
P=x**3-2*x+1 ; Q= x**5-2*x**4+2*x**2-1 ; P1=x**6*y**3-2*x**2*y+1 ; Q1=x**2*y**2-1
display(P,Q,P1,Q1)
```

Le PGCD de deux polynômes se calcule avec la fonction `gcd`et le PPCM avec la fonction `lcm`. Les deux fonctions ne sont pas  réservées uniquement aux polynômes à une seule indéterminée. Elles gèrent aussi les polynômes à plusieurs indéterminées. 

```python
D=gcd(P,Q) ; M=lcm(P,Q) ; display(D,M) 
```

```python
gcdex(P,Q) # PGCD et coefficients de Bezout : retourne un tuple (U,V,D) avec D=PGCD(P,Q) et UP+VQ=D 
```

```python
D1=gcd(P1,Q1) ;  M1=lcm(P1,Q1) ; display(D1,M1) 
```

```python
gcdex(P1,Q1) # gcdex n'a pas de sens pour des polynômes à plusieurs indéterminées car K[X,Y] n'est pas un anneau principal
```

### Construction d'un polynôme

Pour profiter des fonctionnalités propres aux polynômes (degré, coefficients,..), il faut déclarer les polynômes comme des objets de type `Poly` et non comme de simples expressions. On définit une instance de l'objet `Poly` avec la fonction `poly` mais on peut aussi convertir une expression prédéfinie en un objet `Poly` grâce à la méthode `as_poly`. À l'inverse la méthode `as_expr` permet de convertir un objet `Poly` en une expression sympy.    

```python
poly(x**2+sqrt(2)*x+1)  
```

Sans précision, tout «symbole» qui n'est pas explicitement dans l'ensemble $\mathbb Z$ est considéré comme une indéterminée ($\sqrt 2$ en est une dans l'exemple précédent). On peut toutefois préciser l'indéterminée désirée, les autres «symboles» jouant alors le rôle de paramètres. 

```python
poly(x**2+sqrt(2)*x+1,x)  # 'x' est la seule indéterminée
```

Là on constate que SymPy a changé de domaine. C'est le corps de base dans lequel sont pris les coefficient. Le mot clé `EX` est utilisé quand sympy n'arrive pas à décider dans quel domaine il doit se placer. On peut en outre préciser un domaine particulier en utilisant l'option `domaine`. Dans cet exemple le domaine qu'on veut est le corps $\mathbb Q[\sqrt2]$. Il suffit de passer la valeur `QQ(sqt(2))` à l'option `domain`

```python
poly(x**2+sqrt(2)*x+1,domain='QQ(sqrt(2))')
```

Une expression qui contient un `Poly` est coércisée en un objet de type `Poly`

```python
(x-1)*poly(x**2+sqrt(2)*x+1,domain='QQ(sqrt(2))') # 
```

La fonction `poly` retourne toujours une écriture développée sous forme ordonnée par degrés décroissants d'un polynôme 

```python
poly((x-1)**4)
```

```python
P=x-1
Q=poly((x**2+1)**3,x)
P*Q # produit d'une expression et d'un 'Poly', le résultat est un 'Poly'
```

Il peut être important de récupérer le nom de l'indéterminée utilisée dans un `Poly` (pour des besoins de programmation par exemple). On y accède à l'aide de la propriété `gen` de l'objet crée (`gen` pour generator).  

```python
R=poly(x**2+sqrt(2)*x+1,x) ; R.gen
```

Dans le cas de plusieurs indéterminées, utiliser `gens` qui retoune un tuple formé de toutes les indeterminées

```python
R=Poly(x**2+y**2,x,y) 
R.gens
```

On peut convertir un objet de type `Poly` en une expression sympy avec la méthode `as.expr` 

```python
R.as_expr() # convertir un 'Poly' en une 'expression'
```

On peut changer le nom de l'indéterminée dans ce processus de conversion 

```python
R.as_expr(z)
```

### Racines et factorisation 

```python
P=x**3-2*x+1 ; Q=x**5+2*x**3-1 ; R=(x-1)**2*P ; display(P,Q,R)
```

Les racines d'un polynômes sont calculées de façon exacte par la fonction `roots`. La réponse ne peut être complète que si le polynôme est de degré $\leqslant 4$ ou qu'il peut être factorisé en produit de polynômes de degré $\leqslant 4$ sur le domaine en cours (théoriquement une équation algébrique ne peut être résolue par radicaux que si elle est de degré $\leqslant 4$). La sortie de `roots` est un dictionnaire formé de couples `racine:multiplicité` 

```python
roots(P)
```

```python
roots(R)
```

```python
roots(Q) # la réponse est l'ensemble vide car deg Q > 4 et Q est irréductible sur Q
```

```python
roots(R, multiple=true) # s'il est souhaitable d'avoir plutôt la liste des racines, chaque racine est répétée autant de fois que sa multiplicité dans la liste 
```

Pour la factorisation on dispose de la fonction généraliste `factor`. Elle peut être utilisée avec tout type d'expression. Avec des polynômes, la factorisation se fait sur le corps de base $\mathbb Q$. Il faudra l'aider en se plaçant dans un corps qui contient les racines pour obtenir une factorisation complète. Si elle est utilisé avec un `Poly` alors la réponse est un 'Poly' et un 'Poly' est toujours réécrit sous sa forme normale.

```python
factor(P) # factorisanton dans Q[X]
```

On fournit le corps sur lequel doit être effectué la factorisation, ici $\mathbb Q[\sqrt5]$. L'option à utiliser est `extension`

```python
factor(P,extension=sqrt(5)) #  
```

Une option alléchante est prévue : `split`. Elle devrait «en principe» factoriser complètement le polynôme dans $\mathbb C[X]$ mais ne fait pour l'instant que lever une exception (une erreur). Il faudra peut être attendre l’avènement de l'informatique quantique :)

```python
factor(x**3-2*x+1,split=True) # dans un futur proche :)
```

Au lieu de fournir explicitement l'extension à utiliser, on peut définir un alias sur une racine du polynôme et ensuite l'utiliser comme extension.

```python
omega=AlgebraicNumber(RootOf(x**2+x-1,0),alias='omega') # omega est maintenant un alias de la première racine du polynôme fournit à 'RootOf' 
```

```python
P_fact=factor(P,extension=omega) ; P_fact ## factorisation sur le corps Q[omega], une sortie qui peut être beaucoup plus lisible
```

Regardons ce que cela donne si on développe cette expression (avec la méthode `expand`). En principe on devrait retrouver l'expression originale de $P$ (sans $\omega$)

```python
P_fact.expand() # par très souple cette histoire d'extension, d'autre logiciels font mieux, depuis longtemps
```

Sur une simple expression, SymPy ne fait pas le lien avec la signification du symbole $\omega$. Il faudra passer par un `Poly`et préciser que le calcul doit se faire dans le domaine $\mathbb Q[\omega]$.

```python
P_fact.as_poly(extension=omega) # pour retrouver l'expression de P
```

```python
Poly(P_fact,x,extension=omega) # la même chose que l'instruction précédente 
```

```python
Q1=(x-omega)**3
Q2=x**2+omega**2
poly(Q1*Q2,extension=omega) # le calcul des coefficients se fait dans le corps Q[omega] en utilisant la relation omega**2-2*omega+1=0
```

Cette observation permet de faire des calculs dans le corps $\mathbb Q[\omega]$

```python
Poly(omega**6-2*omega+6,x,extension=omega).as_expr()
```

La fonction `factor_list` est plus spécialisée. Elle peut aussi agir sur des objets de type `Poly` et retourne la liste de ses facteurs irréductibles avec leurs évaluations. La factorisation s'effectue dans $\mathbb Q[X]$ à moins de préciser un autre corps d'extension (avec l'option `extension` par exemple).

```python
P=(x-1)**3*(x-2)*(x**3-2*x+1)**2 ; P # on change de polynôme de test
```

```python
factor_list(P,extension=rac) 
```

```python
factor_list(P.as_poly(),extension=rac) #liste des facteurs de dans l'anneau rac[X]
```

```python
factor_list(Q)
```

### Décomposition en éléments simples d'une fraction rationnelle

La fonction `apart` permet de décomposer une fraction rationnelle en éléments simples. La fraction peut être définie comme une simple expression. Les fonctions `together` et `cancel` permettent au contraire de rassembler une expression sous forme d'une fraction.

```python
F=(x**2+1)/(x**3-1) ; G=1/(x-1)+x**2/(1-x**3)
display(F,G)
```

```python
together(G) # ne fait que réécrire le fraction sous forme d'un quotient sans aucun transformation
```

```python
cancel(G) # procède en plus à une simplification des facteurs en commun. 
```

```python
apart(F)
```

Comme pour `factor`, la factorisation du dénominateur se fait sur $\mathbb Q$. La fonction `apart`accepte toutes les options de `factor` et les utilise pour factoriser le dénominateur.

```python
apart(F,extension=(I,sqrt(3))) # les racines du dénominateurs sont dans le corps Q[I,sqrt(3)]
```

Mais aussi de façon plus sophistiquée :

```python
JJ=AlgebraicNumber(RootOf(poly(x**2+x+1),0),alias='j')# on définit un alias 'j' sur la première racine de x²+x+1
```

```python
apart(F,extension=JJ)
```

Une autre option est disponible : `full=True`. Elle déclenche un algorithme particulier pour les calculs.

```python
apart(F, full=True)
```

Pour forcer un calcul complet, on utilise la méthode `doit`

```python
apart(F,full=True).doit()
```

## Définition de matrices 

### Constructeurs généralistes

`Matrix` permet de définir des matrices selon plusieurs variantes de syntaxe. `randMatrix` est une fonction très utile qui construit des matrices dont les coefficients sont choisis aléatoirement. 

```python
A=Matrix([[1,0,1],[2,1,-1],[3,-2,1]]) ; display(A) # liste de listes, chaque liste est une ligne 
```

```python
A=Matrix(((1,0,1),(2,1,-1),(3,-2,1))) ; display(A)  
```

Utiliser des tuples est aussi accepté 

La syntaxe pour construire un vecteur colonne est donc de la forme 

```python
Matrix([[1],[0],[-1]])
```

Mais puisque l'utilisation des vecteurs colonnes est tellement courante, quand l'argument ne contient qu'une seule liste (et non une liste de listes) alors il est considéré comme donnant les coefficients d'une colonne 

```python
Matrix([1,0,-1])
```

Cette facilité se généralise si on fournit à `Matrix` une liste de nom de vecteurs colonnes déjà définis. La matrice construite contient en une seule colonne les vecteurs empilés l'un sous l'autre 

```python
V1=Matrix([1,1,1])
V2=Matrix([2,2,2])
V3=Matrix([3,3,3])
Matrix([V1,V2,V3])
```

Une astuce permet de construire la matrice dont les vecteurs colonnes sont ceux dont les noms sont fournis dans la liste : modifier la taille de la matrice avec les méthode `reshape` et `T` (pour la transposée)

```python
Matrix([V1,V2,V3]).reshape(3,3).T
```

Cela ne fonctionne pas qu'avec des colonnes mais il vaut mieux dans ce cas utiliser les opérations de concaténation ou de construction par blocs pour un meilleur contrôle. 

```python
A1=Matrix(2,2,[1,1,1,1])
A2=Matrix(2,2,[2,2,2,2])
A3=Matrix(2,2,[3,3,3,3])
A4=Matrix(2,2,[4,4,4,4])
Matrix([A1,A2,A3,A4])
```

Il est possible de convertir une matrice en une liste Python. Il suffit d'utiliser pour cela la fonction `list`. C'est bien pratique pour récupérer la liste des coefficients d'une matrice pour les utiliser ensuite dans une équation pae exemple. 

```python
list(A)
```

À ce propos, il ya une variante de syntaxe pour `Matrix` qui accepte dans ses arguments la liste de tous les coefficients à condition de préciser la taille de la matrice. Les coefficients sont alors lus ligne par ligne. 

```python
B=Matrix(3,3,[1,2,3,1,2,3,1,2,3]) ; display(B) # nombre de lignes, nombres de colonne et liste de tous les coefficients
```

```python
x,y,z=symbols('x y z')
M=Matrix(3,3, [1,1,1,x,y,z,x**2,y**2, z**2]) # symboles formels supporté à condition de les définir avec `symbols`
display(M)
```

```python
M.subs(z,10) # crée une nouvelle matrice avec z=10
```

Il y a une autre syntaxe qui peut s'avérer très pratique dans beaucoup de cas : il est possible de fournir la taille de la matrice et une fonction qui définit les coefficients. Le constructeur Python `lambda` est accepté. 

```python
Matrix(9,9, lambda i,j : 1/(i+j+1)) # matrice de Cauchy de taille 9 
```

On peut aussi définir une fonction Python, ou bien un symbole qui désigne le nom d'une fonction,  spécialement pour l'utiliser dans la définition d'une matrice. 

```python
m = Function('m')   # m est maintenant un symbole qui se comporte comme le nom d'une fonction sympy 
def V (n) :         # fonction python qui construit des matrices de Vandermonde
    return Matrix(n,n, lambda i,j : m(j)**i) 
V(5)
```

Pour utiliser des valeurs particulièrs des abscisses on pourrait alors procéder à une substitution :

```python
V(5).subs({m(0):1, m(1):2, m(2):3, m(3):-1, m(4):5}) # subs sur un dictionnaire de valeurs pour m 
```

Pour définir une matrice «circulante»

```python
a=symbols('a:6')
Matrix(6,6, lambda i,j : a[(i+j) %6])
```

Rien n'empêche d'utiliser des instructions conditionnelles dans la fonction qui définit les coefficients 

```python
def f_indice (i,j) :
    if i<j or i==j :
        return a[i]
    else :
        return 0
Matrix(6,6, f_indice)
```

Lors de la construction d'une matrice, SymPy ne fait que placer les coefficients selon l'expression donnée sans presque aucune transformation comme dans l'exemple suivant

```python
Matrix(4,4, lambda i,j : (a[i]**2-a[j]**2)/(a[j]+a[i]))
```

Il faut en général faire une demande explicite de la transformation qu'on veut appliquer aux coefficients avec la méthode `applyfunc`

```python
Matrix(4,4, lambda i,j : (a[i]**2-a[j]**2)/(a[j]+a[i])).applyfunc(simplify)
```

```python
Matrix(4,4, lambda i,j : (a[i]**2-a[j]**2)/(a[j]+a[i])).applyfunc(simplify).applyfunc(lambda x:x**2)
```

Et pour clore cette balade, `randMatrix` permet de construire des matrices dont les coefficients sont générés aléatoirement. Elle prend un argument le nombre de lignes, celui des colonnes et encore deux arguments qui précisent l'intervalle dans lequel seraient pris les coefficients. D'autres syntaxes sont disponibles. 

```python
randMatrix(6,6,-5,5) # taille: (6,6), intervalle des coefficients : [-5,5]
```

Cette fonction est par exemple utile pour générer une matrice $A$ à coefficients dans $\mathbb Z$ dont l'inverse est aussi à coefficients dans $\mathbb Z$, ce qui équivaut à $\det(A)=\pm1$. Une telle matrice peut servir pour créer une matrice semblable à une matrice simple tout en conservant des coefficients dans $\mathbb Z$. 

```python
P=randMatrix(4,4,-2,2) 
while P.det() != 1 :
    P=randMatrix(4,4,-2,2)
display(P, P.inv())
```

## Manipulations de matrices

### Matrices en tant qu'objets itérables

Un objet de type `Matrix` est itérable. Ce qui signifie qu'on peut, par exemple, faire une boucle sur les éléments qu'il contient.     

```python
A=Matrix([[1,0,1],[2,1,-1],[3,-2,1]]) ; display(A)
```

```python
for C in A : # A est itérable, cela peut avoir beaucoup d'intérêt 
    print(f'Ceci est un coefficient de A : {C}')
```

On peut convertir une matrice `A` en un autre objet itérable, une liste par exemple, en récupérant la séquence de ces coefficients avec l'instruction `*A`. Celle-ci ne peut être utilisée que dans un contexte où Python s'attend à trouver une séquence d'éléments. L'instruction '*Obj' fonctionne avec tout objet 'Obj' itérable. 

L'instruction `list(A)` retourne aussi la liste des coefficients de $A$.

```python
list(A)
```

```python
*A
```

```python
[*A] # liste des coefficient de $A$, équivalent à 'list(A)'
```

Cela marche dans le sens inverse : `Matrix` accepte qu'on lui fournisse une matrice à la place de la liste des coefficients (en fait tout objet itérable convient)

## Résolution d'équations 

Sympy dispose de toute une panoplie de résolveurs d'équations. Ceux qui pourront nous concerner ici sont :
+ `solve` : fonction généraliste qui permet de résoudre tout type d'équation, ou système d'équations. Bien que fournissant différents formats pour les solutions selon le type d'équations, la sortie est en général relativement simples à récupérer pour les calculs ultérieurs.
+ `solveset` : une version plus récente plus rigoureuse et avec une interface plus homogène entre les différents type d'équations qui s'efforce de trouver toutes les solutions. La sortie est difficile à exploiter. 
+ `linsolve` :  fonction spécialisée dans la résolution de systèmes linéaires. Elle s'appuie sur `solveset`

`linsesolve` semble indiquée pour les problèmes de l'algèbre linéaire, on lui préférera toutefois la fonction `solve` pour sa souplesse. Le fait que celle-ci retourne un dictionnaire en général, permet de récupérer les solutions par un simple appel à la méthode `subs`


```python
x,y,z,t=symbols('x y z t')
a,b,c,d=symbols('a b c d')
```

`solve` accepte en entrée 
+ une expression 'expr' ou un objet itérable d'expressions (liste, tuple, tableau ou même matrice), par exemple ['expr1', 'expr2',...]. L'équation à résoudre est alors 'expr=0' ou [expr1=0, expr2=0,...]
+ une équation formaté avec le constructeur `Eq` sous la forme `Eq(expr,b)` ou 'E' est une expresion contenant une ou plusieurs inconnue et 'b' est le second membre. L'equation à résoudre est alors 'expr=b'.
+ sans aucune autre indication, tout  symbole utilisé dans le(s) équation(s) est considéré comme inconnu. 
+ on peut préciser les inconnues sous forme d'une séquence non encadrée de symboles ou sous forme d'un objet itérable contenant les noms des inconnues. Dans ce cas les éventuels symboles dans l'equation qui ne sont pas des inconnues sont considérés comme des paramètres.    

La sortie est en général un dictionnaire qui représente les solutions trouvés. Elle peut être vide bien que l'équation admet des solutions (dans certains cas vraiment évidentes). 

```python
solve([a*x+b*y-1, c*x+d*y],x,y) # les inconnues sont x et y. a et b sont des paramètres 
```

```python
solve([a*x+b*y-1, c*x+d*y],(x,y))
```

```python
solve((a*x+b*y-1, c*x+d*y),[x,y])
```

### Exemple 1 : réduction d'une matrice compagne

Sympy fournit la méthode `companion`qui permet de construire la matrice compagne d'un polynôme unitaire donné. 

```python
x=Symbol('x')
P=Poly((x-1)**3*x*(x+1),x)
C=Matrix.companion(P)
display(C)
```

N.B : Il n'est pas difficile de former la matrice compagne d'un polynôme sans utiliser la méthode `companion` comme le montre le code suivant :

```python
Matrix.hstack(banded({-1:(1,1,1,1)}),-Matrix(5,1,[P.nth(k) for k in range(5)]))
```

```python
factor(C.charpoly(x).as_expr())
```

```python
C.diagonalize() # oups, une matrice compagne n'est diagonalisable que si ... 
```

```python
Q,T=C.jordan_form() ; display(Q,T)
```

Sachant $a,b$ et $c$ étant des paramètres *distincts* quelconques, on veut réduire la matrice compagne du polynôme $P=(x-a)(X-b)^2(x-c)^3$ ainsi que sa transposée tout en simplifiant les matrices obtenues pour bien visualiser leur forme générale.

```python
a,b,c,x,y=symbols('a b c x y')
P=poly((x-a)*(x-b)**2*(x-c)**3,x)
C=Matrix.companion(P)
display(P)
display(C)
```

```python
Q,T=C.jordan_form(C)
display(Q,T)
```

```python
simplify(C*Q-Q*T) # vérification, 'C*Q == Q*T' conviendrait aussi
```

Même chose pour la matrice $C^T$ maintenant 

```python
QT,TT=C.T.jordan_form()
QT=QT.applyfunc(simplify)
display(QT,TT)
```

La réponse n'est pas satisfaisante à cause des fractions dans 'QT' qui ne tiennent pas compte du fait que l'un des coefficients $a,b$ ou $c$ peut être nul. Les vecteurs colonnes $V_1,V_2,V_3,V_4,V_5$ de $QT$ sont liées par les relations 
\begin{align}
(C-aI)V_1 &= 0 \\
(C-bI)V_2 &= 0 \\
(C-bI)V_3 &= V_2 \\
(C-cI)V_4 &= 0 \\
(C-cI)V_5 &= V_4 \\
(C-cI)V_6 &= V_5
\end{align}
Ces relations restent inchangées si on multiplie $V_2$ et $V_3$ par un même scalaire et ensuite si on ajoute à $V_3$ un vecteur colinéaire à $V_2$. Une observation similaire peut être faite pour les vecteurs $V_4,V_5$ et $V_6$. Ce qui nous permet de remplacer 'QT' par une matrice plus «générique»

```python
Q_temp=QT[:,:] # on crée un copie de QT pour ne pas avoir à exécuter l'instruction précédente à chaque modification
Q_temp[:,0] *=  a**5 
Q_temp[:,1] *= -b**3
Q_temp[:,2]  = -b**3*Q_temp[:,2]+5/b*Q_temp[:,1]
Q_temp[:,3] *=  c
Q_temp[:,4]  =  c*Q_temp[:,4]+4/c*Q_temp[:,3]
Q_temp[:,5]  =  c*Q_temp[:,5]-10/c**2*Q_temp[:,3]+4/c*Q_temp[:,4]
display(Q_temp)
```

On observe que les colonnes relatives à une même valeur propre $\lambda$ de $C^T$ peuvent être déduites simplement par dérivations successives par rapport à $\lambda$ de la colonne

###  Exemple 3: une fonction plus rapide de diagonalisation 
Écrire, en se basant sur la méthode sympy `eigenvects`, une fonction qui prend en argument une matrice $M$ capable de détecter si $M$ est diagonalisable  et qui retourne dans ce cas une matrice de passage $P$ et une matrice diagonale $D$ telle $M=PDP^{-1}$. 


```python
A=randMatrix(3,3,-5,5) ; display(A)
```

```python
A.eigenvects()
```

```python
def simple_diag (M) : 
    """ 
    adaptée à partir de la méthode 'diagonalize' elle même. 
    Vive le logiciel libre !
    """

    if not M.is_square:
        raise NonSquareMatrixError()
    
    p_cols, diag = [], []
    for val, mult, basis in M.eigenvects() :
        diag   += [val] * mult
        p_cols += basis

    P_cols=Matrix.hstack(*p_cols) # P_cols a pour colonnes tous les vecteurs trouvés par 'eigenvects'
    
    if P_cols.cols < M.cols : # si P_cols a moins de colonnes que A c'est qu'il n'y a pas de base formée de vep. 
        raise MatrixError("Matrice non diagonalisable")
    else : 
        return P_cols, M.diag(*diag)
```

```python
P,D=simple_diag(A) ; display(P,D)
```

### Exemple 6 : évaluation d'un polynôme en une valeur matricielle 

Bizarrement, SymPy ne propose aucune fonction ou méthode pour évaluer un polynôme $P$ en une matrice carrée $M$ (ie calculer $P(M)$). On se propose de combler cette lacune en définissant une méthode qui s'utilise sous la forme `P.eval_matrix(M)` lorsque 'P' a été déclaré comme un `Poly`et 'M' comme une matrice carrée.  

```python
def poly_eval_matrix(self, M): # self sera remplacé par l'instance de l'objet qui fait appel à la méthode. 
    if not M.is_square: # tester si la matrice est carrée
        raise NonSquareMatrixError

    if (len(self.gens) != 1): # teste si le polynôme a été déclaré avec une seule indéterminée 
        raise ValueError("un polynôme a une seule indéterminée est attendu")
    expr = self.as_expr(*P.gens) # convertit le polynôme en une expression
    const, non_const = expr.as_independent(self.gens[0]) # récupère la partie constante et la partie variable du polynôme 

    return non_const.subs(self.gens[0], M).doit() + const * eye(M.rows)

Poly.eval_matrix = poly_eval_matrix # c'est tout bête d'ajouter une méthode à un objet existant 
```

### Exemple 7 : décompostion de Dünford


On veut déterminer la décomposition de Dünford de la matrice

```python
A=Matrix([
[ 6, -5, -2, -1,  3],
[ 5, -4, -2, -1,  3],
[-3,  3,  1,  1, -1],
[-1,  0,  1,  0, -1],
[-2,  1,  1, -1,  0]])
```

```python
A.charpoly().as_expr().factor()
```

On peut utiliser  `jordan_form` pour trigonaliser $A$ et ensuite construire les composantes $D$ et $N$ facilement 

```python
P,T=A.jordan_form()
display(P,T)
```

et récupérer ensuite $D$ et $N$ avec :

```python
D=P*diag(1,1,1,-I,I)*P.inv()
N=A-D
display(D,N)
```

Puisque la décomposition de Dünford est unique et que la matrice de départ $A$ est réelle alors $N$ et $D$ sont en fait réelles. En effet 
$$ A=\overline A\Longrightarrow \overline D- D=N-\overline N $$
On sait que $\overline D$ et $\overline N$ sont des polynômes en $\overline A$ et donc en $A$. Elle commutent donc respectivement avec $D$ et $N$ qui sont elles aussi des polynômes en $A$. On en déduit que $\overline D-D$ est diagonalisable et $N-\overline N$ est nilpotente. Ces deux matrices sont égales donc elles sont nulles. Soit $\overline D=D$ et $\overline N=N$.

Dans notre exemple, il suffit juste de forcer une simplification.

```python
D=simplify(P*diag(1,1,1,-I,I)*P.inv())
N=A-D
display(D,N)
```

```python
D+N == A
```
