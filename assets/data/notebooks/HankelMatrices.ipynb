# Matrices de Hankel d'une matrice carrée 

Une suite $(u_n)_n$ d'éléments d'un espace vectoriel $E$ est dite récurrente linéaire d'ordre $q\in\N^*$ s'il existe  des scalaires   
$a_{q-1},a_{q-2},\cdots,a_1,a_0$ tels que 
$$\forall n\in\mathbb N,\; u_{n+q}+a_{q-1}u_{n+q-1}+\cdots+a_1u_{n+1}+a_0u_n=0$$
Le polynôme  $Q=X^q+a_{q-1}X^{q-1}+\cdots+a_1X+a_0$ est alors dit un polynôme annulateur de $u$.

Si $u$ est récurrente linéaire, l'ensemble de ses polynôme annulateurs est un idéal non nul de $\mathbb K[X]$.  Il est engendré par un unique polynôme annulateur unitaire de $mathbb K[X]$ appelé polynôme minimal de $u$. C'est le polynôme de degré minimal qui annule $u$. Il divise tous les autres polynômes annulateurs de $u$.

Par exemple si $A\in\mathcal M_d(\mathbb K)$, alors la suite $u=(A^n)_n$ est récurrente linéaire et son polynôme minimal est $\pi_A$ le polynôme minimal de $A$.  De même si $V$ est un vecteur colonne alors la suite $v=(A^nV)_n$ est récurrente linéaire et son polynôme minimal est $\pi_{A,V}$ le polynôme minimal de $A$ en $V$.

Étant donné maintenant une suite **scalaire** $\lambda=(\lambda_n)_n$ on appelle matrices de Hankel de $u$ les matrices 
$$ H_n(\lambda)=(\lambda_{i+j-2})_{i,j}=
    \begin{pmatrix} 
        \lambda_0 & \lambda_1 & \cdots & \lambda_n \\
        \lambda_1 & \lambda_2 & \cdots & \lambda_{n+1} \\
        \vdots & \vdots && \vdots \\
        \lambda_{n} & \lambda_{n+1} & \cdots & \lambda_{2n} 
    \end{pmatrix} 
$$
par exemple 
$$H_5(\lambda)=\begin{pmatrix}\lambda_{0} & \lambda_{1} & \lambda_{2} & \lambda_{3} & \lambda_{4} & \lambda_{5}\\\lambda_{1} & \lambda_{2} & \lambda_{3} & \lambda_{4} & \lambda_{5} & \lambda_{6}\\\lambda_{2} & \lambda_{3} & \lambda_{4} & \lambda_{5} & \lambda_{6} & \lambda_{7}\\\lambda_{3} & \lambda_{4} & \lambda_{5} & \lambda_{6} & \lambda_{7} & \lambda_{8}\\\lambda_{4} & \lambda_{5} & \lambda_{6} & \lambda_{7} & \lambda_{8} & \lambda_{9}\\\lambda_{5} & \lambda_{6} & \lambda_{7} & \lambda_{8} & \lambda_{9} & \lambda_{10}\end{pmatrix}$$
(en général une matrices de Hankel est une matrice carrée dont les coefficients sont constants sur chaque ligne transversale)

On démontre qu'une suite scalaire $\lambda=(\lambda_n)_n$ est récurrente linéaire si et seulement s'il existe $p\in\N^*$ tel que 
$$ det(H_{p-1}(u))\ne0 \quad\text{et}\quad (\forall n\geqslant p,\; \det(H_n(u))=0) $$ 
et que dans ce cas il existe un unique vecteur colonne ${}^t(b_{p-1}\;\cdots\;b_1\;b_0)$  tel que 
$$H_p(u)\begin{pmatrix} b_{p-1} \\ b_{p-2} \\ \vdots \\ b_1 \\ b_0 \end{pmatrix} = 0 \quad\text{avec $b_0=1$}$$
De plus dans ces conditions on a  $\pi_\lambda=X^p+b_{1}X^{p-1}+\cdots+b_{p-2}X+b_{p-1}$.

Si $A$ est une matrice carrée alors pour tout couple $(V,W)$ de vecteurs colonnes la suite scalaire $(^{t}VA^nW)_n$ est récurrente linéaire et son polynôme minimal, noté $\pi_{A,V,W}$, divise le polynôme minimal $\pi_A$ de $A$.

En plus du résultat usuel selon lequel il existe un vecteur $W$ tel que $\pi_A=\pi_{A,W}$, on démontrer que pour tout vecteur $V$ il existe $W$ tel que $\pi_{V,A,W}=\pi_{A,W}.

On se propose dans cette planche d'expérimenter sur la fréquence à laquelle on obtient des vecteurs $V$ et $W$ tels que $\pi_A=\pi_{V,A,W}$ lorsque on les choisit aléatoirement.

Une méthode de l'objet `Matrix` qui calcule le polynôme minimal 

```python
x=Symbol('x')
def eval_matrix_minpoly(self, gen=x) :
    nbr_cols=min(self.cols, self.rank()+1)
    M_temp=Matrix([list(self**k) for k in range(nbr_cols+1)]).T
    M_temp=M_temp.rref()[0]
    deg_minpoly=M_temp.rank()
    return Poly(gen**deg_minpoly-sum(M_temp[k,deg_minpoly]*gen**k for k in range(deg_minpoly)))
Matrix.minpoly = eval_matrix_minpoly 
SparseMatrix.minpoly=eval_matrix_minpoly
```

Définition de la fonction `hankel` qui prend en argument la taille 'n' de la matrice de Hankel à construire, le nom optionnel d'une fonction (le nom de la suite) et un argument optionnel sous la forme `sequence=liste` auquel on doit passer une liste de termes de la suite en nombre suffisant pour construire la matrice de Hankel. 

Le paramètre 'n' est obligatoire. Si un deuxième paramètre est présent il sera considéré comme un nom de fonction et sera utilisé pour définir les termes de la suite. Une expression `lambda` est acceptée pour ce paramètre. Si un argument `sequence=liste` est présent, le deuxième paramètre sera ignoré et c'est la liste `liste` (prédéfinie ou donnée explicitement ) qui fournira les termes de la suite. Dans ce cas `liste`  devrait contenir *au moins* $2n+1$ pour former la matrice.

```python
def hankel(n, function=None, sequence:list=[]) :
    if len(sequence) > 0 :
        if len(sequence) < 2*n+1 :
            raise ValueError("sequence ne contient pas assez de termes")
        else :
            function = lambda i : sequence[i]
    return Matrix(n+1,n+1, lambda i,j : function(i+j))
```

```python
u=symbols('u',cls=Function)
hankel(4,u) # Matrice de Hankel de la suite symbolique 'u' 
```

```python
x=symbols('x')
hankel(4,lambda m:m**2*x**m) # Matrice de Hankel à l'ordre 4 de la suite (n²x^n)_n
```

```python
lamb=symbols('lambda:11')
hankel(5,sequence=lamb) # Matrice de Hankel puisant ses coefficients dans le tuple de symboles 'lamb' 
```

Fonction `conv_puiss` qui calcule la matrice de Hankel de deux vecteurs colonnes $V$ et $W$ en une matrice $A$ à un ordre indiqué. Elle prend :
+ trois arguments obligatoires : 'A', 'V' et 'W', respectivement une matrice carrée et deux vecteurs colonnes ;
+ un argument optionnel qui indique le nombre de termes de la suite $({}^tVA^nW)_n$ qui seront calculés. En son absence, $2d+1$ termes seront calculés, $d$ étant l'ordre de la matrice. Juste le nombre de termes nécessaire pour construire la matrice de Hankel d'ordre $d$ de la suite en question.  

```python
def conv_puiss(A,V,W,d=2*(A.rows+1)) :
    U=W 
    termes=[]
    for _ in range(d+1) :
        U = A*U 
        termes.append(*V.T*U)
    return termes 
```

Fonction qui étudie la suite des matrices de Hankel d'une matrice 'A' en deux vecteurs 'V' et 'W' et retourne le polynôme minimal de $A$ en $(V,W)$. La fonction retourne `None` si la suite $({}^tVA^nW)_n$ est partout nulle ou si tous les déterminants de Hankel sont nuls. 

```python
def hankel_minpoly(A,V,W) :
    L=conv_puiss(A,V,W)
    L_det=[hankel(r,sequence=L).det() for r in range(A.rows+1)] 
    r=A.rows
    while L_det[r] == 0 and r>0:
        r -= 1
    # r est l'indice du dernier élément non nul dans L_det
    if r == 0 :
        return None 
    else :
        r += 1 # r est l'indice du dernier élément non nul dans L_det
        U=hankel(r,sequence=L).nullspace()
        return (sum(U[0][i,0]*x**i for i in range(r+1))).as_poly(x)
```

```python
display(Math('\\pi_A=%s \\qquad \\pi_{\\mathrm{Hankel},A}=%s'
              %(latex(A.minpoly().as_expr()),latex(hankel_minpoly(A,V,W).as_expr()))))
```

Le block de code suivant forme une matrice aléatoire et lance ensuite une boucle qui choisit à chaque fois deux vecteurs $V$ et $W$ dont les coefficients sont pris aléatoirement dans l'intervalle $\llbracket-\mathrm{interval}, \mathrm{interval}\rrbracket$ et comptabilise les cas où $\deg\pi_{A,V,W}\ne\chi_A$

```python
N=1000
interval=1
cas = 0 
A=randMatrix(5,5,-5,5)
for _ in range(N) :
    V=randMatrix(5,1,-interval,interval)
    W=randMatrix(5,1,-interval,interval)
    P=hankel_minpoly(A,V,W)
    if  P==None or P.degree() < 5 :
        cas += 1
print(f"le nombre de cas trouvé est {cas} sur {N} lorsque l'échantillon comporte {2*interval+1} éléments")
```

```python
N=1000
interval=5
cas = 0 
A=randMatrix(5,5,-5,5)
s=A.minpoly().degree()
for _ in range(N) :
    V=randMatrix(5,1,-interval,interval)
    W=randMatrix(5,1,-interval,interval)
    P=hankel_minpoly(A,V,W)
    if  P==None or P.degree() < s :
        cas += 1
print(f"le nombre de cas trouvé est {cas} sur {N} lorsque l'échantillon comporte {2*interval+1} éléments")
```

Dans le bloc suivant on génère une matrice $A$ d'ordre $6$ diagonalisable avec un polynôme minimal de degré plus petit  ($\pi_A=(X-1)(X+1)(X-2)$) et on refait les tests précédents.

```python
Q=randMatrix(6,6,-1,1)
while Q.det() != 1 :
    Q=randMatrix(6,6,-1,1)
A=Q*diag(1,1,1,-1,2,2)*Q.inv()
R=A.minpoly()
display(A,R.as_expr().factor())
```

```python
N=1000
interval=1
cas = 0 
for _ in range(N) :
    V=randMatrix(6,1,-interval,interval)
    W=randMatrix(6,1,-interval,interval)
    P=hankel_minpoly(A,V,W)
    if  P==None or P.degree() < R.degree() :
        cas += 1
print(f"le nombre de cas trouvé est {cas} sur {N} lorsque l'échantillon comporte {2*interval+1} éléments")
```
